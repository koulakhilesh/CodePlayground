# London's Reservoirs Meet a Record-Hot Summer (2026)

**Data:** 1 Jan 1989 – 31 Jul 2026  ·  **Notebook last run:** 2026-08-15

A short follow-up to [the first reservoir piece](https://koulakhilesh.github.io/writing/london-reservoirs/), which plotted 37 years of daily levels and found that the drought everyone *remembers* (2022) wasn't the one that ran London driest. That post stopped at 31 May 2026. Six weeks later the summer arrived — and the Met Office now says a [record-warm UK summer is increasingly likely](https://www.metoffice.gov.uk/blog/2026/warmest-uk-summer-on-record-increasingly-likely-as-temperatures-stay-well-above-average). This notebook adds June and July 2026 and asks a single question: **how are the reservoirs responding?**

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

pd.set_option("display.max_columns", None)

GROUPS = {"lower_lee_group": "Lower Lee", "lower_thames_group": "Lower Thames"}
COLORS = {"Lower Lee": "#2a9d8f", "Lower Thames": "#e76f51"}
HILITE = "#c1121f"  # 2026 accent
```

## 1. Load and clean

Same file as before, extended to 31 July 2026 (plus a couple of trailing blank rows to drop). The three familiar gotchas still apply: stray spaces in the headers, mixed date formats, and a few non-numeric cells coerced to `NaN`.


```python
raw = pd.read_csv("../data/london_reservoir_levels-july-2026.csv")
raw.columns = raw.columns.str.strip()                       # gotcha #1: header spaces

df = raw[["date", *GROUPS]].copy()
df["date"] = pd.to_datetime(df["date"], format="mixed", dayfirst=True, errors="coerce")
df = df.dropna(subset=["date"])                             # drop trailing blank rows

for col in GROUPS:                                          # gotcha #3: n/a, ---
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["doy"] = df["date"].dt.dayofyear
df = df.sort_values("date").reset_index(drop=True)

print(f"Rows       : {len(df):,}")
print(f"Date range : {df['date'].min():%Y-%m-%d} -> {df['date'].max():%Y-%m-%d}")
latest = df.iloc[-1]
print(f"Latest read: Lower Lee {latest['lower_lee_group']:.0f}%  |  "
      f"Lower Thames {latest['lower_thames_group']:.0f}%  ({latest['date']:%d %b %Y})")
df.tail(3)
```

## 2. The whole record, with 2026 picked out

Every daily reading since 1989, both groups, with the 2026 window shaded so the latest plunge is easy to find against four decades of sawtooth.


```python
long = df.melt("date", value_vars=list(GROUPS), var_name="group", value_name="pct")
long["group"] = long["group"].map(GROUPS)

fig = px.line(
    long, x="date", y="pct", color="group",
    color_discrete_map=COLORS,
    labels={"pct": "% of capacity", "date": "", "group": ""},
    title="London reservoir levels, 1989–2026 (daily % of capacity)",
)
# "full" on the left so it can't collide with the 2026 label on the right.
fig.add_hline(y=100, line_dash="dot", line_color="#999",
              annotation_text="full", annotation_position="top left")
fig.add_vrect(x0="2026-01-01", x1=str(df["date"].max().date()),
              fillcolor="rgba(193,18,31,0.10)", line_width=0)
fig.add_annotation(x=str(df["date"].max().date()), y=0.98,
                   xref="x", yref="y domain", xanchor="right", yanchor="top",
                   showarrow=False, text="2026", font=dict(color=HILITE, size=12))
fig.update_layout(template="plotly_white", hovermode="x unified", legend_title="")
fig_daily = fig
fig.show()
```

## 3. 2026 against the 37-year normal band

The most honest way to judge a year is to draw it on top of the **normal band**: the grey envelope is the full day-by-day min–max of every *other* year (1989–2025), and the dashed line is the typical year. Where does 2026 sit — and how fast is it falling?


```python
focus_year = 2026
fig = make_subplots(
    rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.09,
    subplot_titles=[f"{GROUPS[c]}" for c in GROUPS],
)
for r, (col, name) in enumerate(GROUPS.items(), start=1):
    hist = df[df["year"] != focus_year].groupby("doy")[col].agg(["min", "max", "mean"])
    this = df[df["year"] == focus_year].set_index("doy")[col].sort_index()
    fig.add_trace(go.Scatter(x=hist.index, y=hist["max"], line=dict(width=0),
                             showlegend=False, hoverinfo="skip"), row=r, col=1)
    fig.add_trace(go.Scatter(x=hist.index, y=hist["min"], fill="tonexty",
                             fillcolor="rgba(150,150,150,0.25)", line=dict(width=0),
                             name="1989–2025 range", showlegend=(r == 1),
                             hoverinfo="skip"), row=r, col=1)
    fig.add_trace(go.Scatter(x=hist.index, y=hist["mean"], line=dict(color="#888", dash="dash"),
                             name="typical year", showlegend=(r == 1)), row=r, col=1)
    fig.add_trace(go.Scatter(x=this.index, y=this.values, line=dict(color=COLORS[name], width=3),
                             name=f"{name} 2026"), row=r, col=1)

fig.update_layout(template="plotly_white", hovermode="x unified", height=640,
                  title="London reservoirs: 2026 against the 37-year normal band")
fig.update_xaxes(title_text="day of year", row=2, col=1)
fig.update_yaxes(title_text="% of capacity")
fig_band = fig
fig.show()

# Quantify what the band shows: winter peak -> end of May -> latest, and today's standing.
latest_doy = int(df[df["year"] == focus_year]["doy"].max())
for col, name in GROUPS.items():
    hist_doy = df[(df["year"] != focus_year) & (df["doy"] == latest_doy)][col].dropna()
    now = df[(df["year"] == focus_year) & (df["doy"] == latest_doy)][col].iloc[0]
    peak = df[(df["year"] == focus_year) & (df["doy"] <= 120)][col].max()        # winter/spring high
    may_end = df[(df["year"] == focus_year) & (df["month"] == 5)][col].iloc[-1]   # 31 May
    pctile = (hist_doy < now).mean() * 100
    standing = ("record low for the date" if now <= hist_doy.min()
                else f"{pctile:.0f}th percentile for the date")
    print(f"{name:13s}: winter peak {peak:.0f}%  |  end-May {may_end:.0f}%  ->  "
          f"{now:.0f}% on {df['date'].max():%d %b}  ({standing}; "
          f"37-yr range for the day {hist_doy.min():.0f}-{hist_doy.max():.0f}%)")
```

## 4. How fast did the summer draw down?

A hot, dry summer shows up as *speed*: how much the reservoirs fall between late spring and mid-summer. For every year I take the average level in the **last week of May** and subtract the average in the **last week of July** — the size of the June–July drawdown — then rank the years. A bigger number means a faster summer fall.


```python
end_may = (df["month"] == 5) & (df["date"].dt.day >= 25)
end_jul = (df["month"] == 7) & (df["date"].dt.day >= 25)

rows = []
for col, name in GROUPS.items():
    may = df[end_may].groupby("year")[col].mean()
    jul = df[end_jul].groupby("year")[col].mean()
    drop = (may - jul).dropna()
    for y, v in drop.items():
        rows.append({"year": int(y), "group": name, "drop": v})
dd = pd.DataFrame(rows)

for name in GROUPS.values():
    g = dd[dd["group"] == name].sort_values("drop", ascending=False).reset_index(drop=True)
    rank = int(g.index[g["year"] == 2026][0]) + 1
    val = g.loc[g["year"] == 2026, "drop"].iloc[0]
    print(f"{name:13s}: 2026 May->July fall = {val:4.1f} pts  (rank {rank} of {len(g)} summers; "
          f"steepest ever = {g['drop'].iloc[0]:.1f} in {int(g['year'].iloc[0])})")

th = dd[dd["group"] == "Lower Thames"].sort_values("drop").tail(12)
bar_colors = [HILITE if y == 2026 else COLORS["Lower Thames"] for y in th["year"]]
fig = go.Figure(go.Bar(
    x=th["drop"], y=th["year"].astype(str), orientation="h",
    marker_color=bar_colors,
    hovertemplate="%{y}: %{x:.1f} pts<extra></extra>",
))
fig.update_layout(
    template="plotly_white",
    title="Lower Thames: steepest May-to-July drawdowns (2026 in red)",
    xaxis_title="drop in % of capacity, late May -> late July",
    yaxis_title="", margin=dict(l=60, r=20, t=60, b=40),
)
fig_drop = fig
fig.show()
```

## 5. Takeaways

- **The Lower Thames is riding the floor.** At 72% on 31 July 2026 it's sitting on the bottom edge of its 37-year normal band — matching the lowest level ever recorded for that date. The Lower Lee, at 82%, is low but less extreme (around the 24th percentile for the day).
- **Fast, but not the fastest.** The June–July fall was steep — about 17 points on the Thames (5th-steepest of 38 summers) and 13 on the Lee (7th) — yet 2022 (Thames, ~24 points) and 2018 (Lee, ~22 points) both fell faster. A record-hot summer did **not** produce a record-fast drawdown.
- **It started from a good place.** Both groups were near the top of the band in spring (the Lee touched 100%), so even a record-low-for-the-date July hasn't tipped into crisis — the reservoirs are spending a full winter's savings, not running on empty.
- **The autumn low is the real test.** July isn't the usual trough — that comes in September–October. If the Thames keeps hugging the floor, the seasonal minimum is the number to watch against the deep droughts of 1996–97.

## 6. Save the interactive charts

Export the three figures as standalone interactive HTML (Plotly from CDN) for embedding.


```python
from pathlib import Path

EXPORT_DIR = Path("exports")
EXPORT_DIR.mkdir(exist_ok=True)

VIEWPORT_META = '<meta name="viewport" content="width=device-width, initial-scale=1" />'


def inject_viewport(path: Path) -> None:
    html = path.read_text()
    if 'name="viewport"' in html:
        return
    html = html.replace('<meta charset="utf-8" />', '<meta charset="utf-8" />' + VIEWPORT_META, 1)
    path.write_text(html)


html_config = {"responsive": True, "displaylogo": False}
charts = {
    "daily-2026": fig_daily,
    "normal-band-2026": fig_band,
    "summer-drawdown": fig_drop,
}
for slug, figure in charts.items():
    figure.update_layout(autosize=True, margin=dict(l=60, r=20, t=60, b=40))
    out = EXPORT_DIR / f"{slug}.html"
    figure.write_html(out, include_plotlyjs="cdn", full_html=True, config=html_config)
    inject_viewport(out)
    print(f"wrote {out}")
```
